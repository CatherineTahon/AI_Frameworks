## Data Processing

```python
data.loc[[72]]
```

```python
#Give a title for the ticket
data.columns = ['Time', 'Title', 'Description', 'Class']
data.loc[:0]
```

```python
data.loc[[72]]
```

```python
#What to do with the missing ticket?
data = data.dropna(subset=['Description'])
data = data.drop(columns='Time')
data.info()
```

```python
#join the title with the ticket description
data['Title'] = data['Title'].fillna(data['Description'])
data.info()
```

```python
#Concat because we need one tekst and one target
data['TitleDesc'] = data['Title'] + ' ' + data['Description']
data
```

```python
data.loc[[72]]
```

```python
#Drop possible columns 
data = data.drop(columns=['Title','Description'])
data
```

```python
#rename again the column to: Description
data = data.rename(columns={data.columns[1]: 'Description'})
data
```

```python
X = data['Description'] 
y = data['Class']
```

Split your data already to be sure that no data likeage will happens

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state = 4)
```

```python
train = pd.concat([X_train, y_train], axis =1)
train.to_csv('train_tickets.csv')
```

```python
test = pd.concat([X_test, y_test], axis =1)
test.to_csv('test_tickets.csv')
```
